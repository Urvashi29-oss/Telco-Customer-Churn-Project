```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report,confusion_matrix,roc_auc_score

```

```python

##TotalCharges sometimes has blank strings - clean it,
telco['TotalCharges'] = pd.to_numeric(telco['TotalCharges'],errors='coerce')
telco = telco.dropna(subset=['TotalCharges'])
```

```python
telco = telco.drop(columns=['customerID'])
```

```python
#Encode Target
telco.columns = telco.columns.str.strip()
telco['Churn'] = telco['Churn'].map({'Yes': 1,'No': 0})
telco_encoded = pd.get_dummies(telco, drop_first=True)
```

```python
print(telco_encoded.columns.tolist())
```

```python
#telco_encoded = telco_encoded.dropna(subset=['Churn'])
```

```python
#print(Y.isna().sum())
```

```python

print(telco['Churn'].unique())
```

```python
#Split features and Targets
X = telco_encoded.drop(['Churn','customerID'],axis=1,errors='ignore')
Y = telco_encoded['Churn']

X_train,X_test,Y_train,Y_test = train_test_split(
    X,Y,test_size=0.2,random_state=42,stratify=Y
)
```

```python
#Trainthemodel
model = LogisticRegression(max_iter=1000)
model.fit(X_train,Y_train)
```

```python
#Evaluate
Y_pred = model.predict(X_test)
Y_proba = model.predict_proba(X_test)[:,1]

print(classification_report(Y_test,Y_pred))
print("ROC-AUC:",roc_auc_score(Y_test,Y_proba))
```

```python
#See which features matter most
coefs = pd.Series(model.coef_[0],index=X.columns).sort_values()
print(coefs.head(10)) #strongest churn-reducing features
print(coefs.tail(10)) #strongest churn-driving features


```
